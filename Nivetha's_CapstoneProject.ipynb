# **Feature Engineering**

```python
# Created a new Feature called Growth category

def growth_category(x):
    if x < 0:
        return "Declining"
    elif x < 5:
        return "Stable"
    elif x < 15:
        return "Growing"
    else:
        return "High Growth"

df["Growth_Category"] = df["Growth_Percent_24-34"].apply(growth_category)
```

The Purpose of creating the feature called Growth category is to Simplify the Workforce Segamnetation. Make it easier to identify Occupation that need Different statergic Attention.

```python
# Undestanding the Growth_category

df
```

```python
#Created the another feature called Skill_level
#Categorize skill importance


def skill_level(x):
    if x < 2:
        return "Low"
    elif x < 3:
        return "Moderate"
    elif x < 4:
        return "High"
    else:
        return "Critical"

df["Skill_Level"] = df["Skill_Score"].apply(skill_level)
```

I created a new feature called Skill_Level from the existing Skill_Score. I converted the numerical skill score into four categories—Low, Moderate, High, and Critical—using defined score ranges. This makes the skill information easier to interpret and use in categorical analysis

```python

#created the another Feature called Employment_size

#Categorize current workforce size

def employment_size(x):
    if x < 50000:
        return "Small"
    elif x < 200000:
        return "Medium"
    else:
        return "Large"

df["Employment_Size"] = df["Employment_2024"].apply(employment_size)
```

I created an Employment_Size feature from the 2024 employment value. I categorized occupations into Small, Medium, and Large employment groups using predefined thresholds. This helps compare occupations based on their current workforce size.

```python
#created the Another Feature Called Future Demand Index

#Combine growth and skill into one indicator

df["Future_Demand_Index"] = (
    df["Growth_Percent_24-34"] * df["Skill_Score"]
)
```



I created a derived Future Demand Index by combining projected occupational growth with skill score. The purpose was to create a single indicator that considers both future growth and skill importance. It is a project-derived metric rather than an externally published measure, so I use it as an analytical indicator rather than actual future demand.

# **Stage: 3**

#**Statistical Analysis**

```python
occupation_df = (
    df[
        [
            "Occupation",
            "Occupation_Code",
            "Employment_2024",
            "Employment_2034",
            "Employment_Change_24-34",
            "Growth_Percent_24-34"
        ]
    ]
    .drop_duplicates(subset="Occupation_Code")
    .reset_index(drop=True)
)

print(occupation_df.shape)
```

```python
#Statistical Visualization — Mean, Median & Mode

growth = occupation_df["Growth_Percent_24-34"].dropna()

# Statistical values
mean_growth = growth.mean()
median_growth = growth.median()

# Mode
mode_growth = growth.mode()[0]

# Create histogram bins
bins = 30
counts, bin_edges = np.histogram(growth, bins=bins)

# Find the most frequent histogram bin
mode_bin = np.argmax(counts)

mode_bin_center = (
    bin_edges[mode_bin] + bin_edges[mode_bin + 1]
) / 2

# Plot
plt.figure(figsize=(13, 7))

sns.histplot(
    growth,
    bins=bins,
    kde=True,
    stat="count",
    alpha=0.65,
    edgecolor="white",
    linewidth=0.8
)

# Mean
plt.axvline(
    mean_growth,
    linestyle="--",
    linewidth=2.5,
    label=f"Mean = {mean_growth:.2f}%"
)

# Median
plt.axvline(
    median_growth,
    linestyle="-",
    linewidth=2.5,
    label=f"Median = {median_growth:.2f}%"
)

# Mode / highest-frequency region
plt.axvline(
    mode_bin_center,
    linestyle=":",
    linewidth=2.5,
    label=f"Modal Region ≈ {mode_bin_center:.2f}%"
)

plt.title(
    "Distribution of Projected Employment Growth Across Occupations",
    fontsize=17,
    fontweight="bold",
    pad=18
)

plt.xlabel(
    "Projected Employment Growth (2024–2034)",
    fontsize=12
)

plt.ylabel(
    "Number of Occupations",
    fontsize=12
)

plt.legend(
    frameon=False,
    fontsize=11
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

sns.despine()

plt.tight_layout()
plt.show()
```

```python
print("Mean:", round(mean_growth, 2))
print("Median:", round(median_growth, 2))
print("Mode:", round(mode_growth, 2))
```

**Interpretation**

The histogram shows the distribution of projected employment growth across occupations, with the mean, median, and modal region highlighted. Comparing these measures helps identify the central tendency and whether employment growth is symmetrically distributed or influenced by occupations with unusually high or low projected growth.

**Key insight**

The relationship between the mean, median, and mode indicates the underlying shape of the employment-growth distribution. A noticeable separation between these measures suggests skewness, while close alignment indicates a more concentrated and symmetric growth pattern.

# **1. High-Growth Occupations: Current vs Future Employment**

**Type: Multivariate dumbbell-style visualization**

```python
high_growth = occupation_df[
    occupation_df["Growth_Percent_24-34"] >=
    occupation_df["Growth_Percent_24-34"].quantile(0.75)
].copy()

high_growth = high_growth.sort_values(
    "Employment_Change_24-34",
    ascending=False
).head(15)

plt.figure(figsize=(10, 6))

for i, (_, row) in enumerate(high_growth.iterrows()):
    plt.plot(
        [row["Employment_2024"], row["Employment_2034"]],
        [i, i],
        linewidth=3,
        alpha=0.65
    )

    plt.scatter(
        row["Employment_2024"],
        i,
        s=80,
        zorder=3
    )

    plt.scatter(
        row["Employment_2034"],
        i,
        s=100,
        zorder=3
    )

plt.yticks(
    range(len(high_growth)),
    high_growth["Occupation"]
)

plt.xlabel("Employment")
plt.ylabel("Occupation")

plt.title(
    "Future Workforce Expansion Across High-Growth Occupations",
    fontsize=16,
    fontweight="bold"
)

plt.grid(axis="x", alpha=0.2)
sns.despine()
plt.tight_layout()
plt.show()
```

**Current → Future Workforce Expansion**

**Interpretation:**
The chart shows how employment is expected to change from 2024 to 2034 among high-growth occupations. The distance between the current and future employment points represents the expected workforce expansion.

**Key insight:**
The occupations with the largest absolute employment increase will potentially require the greatest amount of future workforce capacity.

**Observation:**
Look for the occupations with the longest expansion distance between 2024 and 2034. These are more important for workforce planning than occupations with high percentage growth but relatively small employment bases.

**Critical Workforce Exposure**

**Type: Bivariate ranking**

```python
critical = occupation_df.copy()

critical["Workforce_Exposure"] = (
    critical["Employment_2024"] *
    abs(critical["Growth_Percent_24-34"])
)

critical = critical.sort_values(
    "Workforce_Exposure",
    ascending=False
).head(15)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=critical,
    y="Occupation",
    x="Workforce_Exposure"
)

plt.title(
    "Occupations with Highest Workforce Exposure",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Workforce Exposure Index")
plt.ylabel("Occupation")

plt.grid(axis="x", alpha=0.2)
sns.despine()
plt.tight_layout()
plt.show()
```

**Workforce Exposure**

**Interpretation:**
The Workforce Exposure Index combines the size of an occupation's current workforce with the magnitude of its projected growth or decline.

**Key insight:**
An occupation with a large workforce can have high workforce exposure even when its percentage growth is relatively moderate.

**Observation:**
The highest-ranked occupations represent areas where changes in employment could affect a substantial workforce population and therefore deserve closer planning attention.

# **4.Skill Demand Across Patterns**

```python
skill_pattern_df = (
    df.groupby(
        ["Growth_Category", "Skill"],
        as_index=False
    )["Skill_Score"]
    .mean()
)
```

**Skill Demand Heatmap**

**Type: Multivariate heatmap **

```python
top_skills = (
    skill_pattern_df.groupby("Skill")["Skill_Score"]
    .mean()
    .nlargest(20)
    .index
)

heatmap_data = (
    skill_pattern_df[
        skill_pattern_df["Skill"].isin(top_skills)
    ]
    .pivot(
        index="Skill",
        columns="Growth_Category",
        values="Skill_Score"
    )
)

plt.figure(figsize=(12, 9))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".1f",
    linewidths=0.5
)

plt.title(
    "Skill Demand Across Employment Growth Patterns",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Growth Pattern")
plt.ylabel("Skill")

plt.tight_layout()
plt.show()
```

**Interpretation:**
The heatmap compares average skill scores across employment-growth patterns. Darker/higher-value cells indicate stronger skill demand within a particular growth pattern.

**Key insight:**
The visualization identifies skills whose importance is concentrated in growing and high-growth occupations, helping distinguish future-oriented skills from skills associated mainly with stable or declining occupations.

**Observation:**
Look for skills with consistently high scores in the High Growth column. Skills that remain strong across several growth patterns are broadly transferable, while skills concentrated in one pattern are more specialized.

**Skill Demand Momentum**

**Type: Bivariate ranking**

```python
skill_pivot = (
    skill_pattern_df
    .pivot(
        index="Skill",
        columns="Growth_Category",
        values="Skill_Score"
    )
)

skill_pivot["Skill_Momentum"] = (
    skill_pivot["High Growth"] -
    skill_pivot["Declining"]
)

momentum = (
    skill_pivot["Skill_Momentum"]
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=momentum,
    y="Skill",
    x="Skill_Momentum"
)

plt.axvline(0, linestyle="--", linewidth=1.5)

plt.title(
    "Skills with the Strongest Growth-Pattern Momentum",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Skill Demand Difference: High Growth − Declining")
plt.ylabel("Skill")

plt.grid(axis="x", alpha=0.2)
sns.despine()
plt.tight_layout()
plt.show()
```

**Skill Demand Momentum**

**Interpretation:**
Skill Momentum measures the difference between a skill's average demand in High Growth occupations and its demand in Declining occupations.

**Key insight:**
A positive momentum score indicates that the skill is more strongly associated with high-growth occupations and can therefore be considered a potential future-oriented skill.

**Observation:**
The skills at the top of the chart have the largest positive difference between High Growth and Declining occupations. These are strong candidates for upskilling and workforce-development programs.
